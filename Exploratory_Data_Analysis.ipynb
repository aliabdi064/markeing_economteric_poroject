# Exploratory Data Analysis of Advertising Data

## Introduction

In this notebook, we will explore the `advertising.csv` dataset. The dataset contains information about advertising spending across different channels (TV, Radio, Newspaper) and the corresponding sales. Our goal is to understand the relationship between the advertising channels and sales, and to identify which channels are most effective in driving sales.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
data = pd.read_csv('advertising.csv')

# Display the first 5 rows
data.head()
```

## Data Inspection

```python
# Get information about the dataset
data.info()
```

```python
# Check for missing values
data.isnull().sum()
```

## Descriptive Statistics

```python
data.describe()
```

## Univariate Analysis

```python
# Histograms
data.hist(bins=20, figsize=(14,10))
plt.show()
```

```python
# Box plots
plt.figure(figsize=(14,10))
sns.boxplot(data=data)
plt.show()
```

## Bivariate Analysis

```python
# Scatter plots
sns.pairplot(data, x_vars=['TV', 'Radio', 'Newspaper'], y_vars='Sales', height=4, aspect=1, kind='scatter')
plt.show()
```

```python
# Correlation matrix and heatmap
corr_matrix = data.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.show()
```

## Statistical Tests

```python
# Correlation test
from scipy.stats import pearsonr

for col in ['TV', 'Radio', 'Newspaper']:
    corr, p_value = pearsonr(data[col], data['Sales'])
    print(f'Correlation between {col} and Sales: {corr:.3f}, p-value: {p_value:.3f}')
```

## Conclusion

Based on the exploratory data analysis, we can conclude the following:

- **TV advertising** has the strongest positive correlation with sales.
- **Radio advertising** also has a positive correlation with sales, but weaker than TV.
- **Newspaper advertising** has the weakest correlation with sales.
- The relationships between TV/Radio and Sales appear to be linear.
- The statistical tests confirm that the correlations between TV/Radio and Sales are statistically significant.
